```python
# Feature selection
X = df[['Annual Income (k$)', 'Spending Score (1-100)']]
```

```python
# Train K-Means Model
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X)
```

```python
df.head()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(df["Annual Income (k$)"],df["Spending Score (1-100)"],c=df["Cluster"],cmap="viridis",s=80)
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("Customer Segmentation using K-Means")
plt.show()
```

```python
df.groupby("Cluster")[["Annual Income (k$)", "Spending Score (1-100)"]].mean()
```

```python
df["Customer Type"] = df["Cluster"].map({0: "Budget-shoppers",1: "High-spenders",2: "Occasional buyers"})
df[["Annual Income (k$)", "Spending Score (1-100)", "Customer Type"]].head(10)
```

```python
print(df["Customer Type"].value_counts())
```

```python
# user input
income = float(input("Enter Annual Income (k$): "))
score = float(input("Enter Spending Score (1-100): "))
customer = pd.DataFrame([[income, score]], columns=X.columns)
cluster = kmeans.predict(customer)[0]
if cluster == 0:
    print("\nBudget-shoppers")
elif cluster == 1:
    print("\nHigh-spenders")
else:
    print("\nOccasional buyers")
```
